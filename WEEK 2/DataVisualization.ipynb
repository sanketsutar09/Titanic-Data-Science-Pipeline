```python
#Importing the required libraries and reading the dataset
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("D:/INTERNSHIP/TitanicDataset_cleaned.csv")
df
```

```python
plt.figure(figsize=(10,6))
sns.countplot(x="Pclass", data=df, palette="Set2", linewidth=2)
plt.title("Distribution of Passengers Across Ticket Classes")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.legend(title="Passenger Class", labels=["1st Class", "2nd Class", "3rd Class"])
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(x="Sex", hue="Survived", data=df, palette="Set1", linewidth=2)
plt.title("Survival Rate by Gender", fontsize=18, weight="bold")
plt.xlabel("Gender", fontsize=13)
plt.ylabel("Passenger Count", fontsize=13)
plt.legend(title="Survived", labels=["No", "Yes"])
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(x="Pclass", hue="Survived", data=df, palette="Set2", linewidth=2)
plt.title("Survival Rate by Passenger Class", fontsize=18, weight="bold")
plt.xlabel("Passenger Class", fontsize=13)
plt.ylabel("Passenger Count", fontsize=13)
plt.legend(title="Survived", labels=["No", "Yes"])
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data=df, x="Age", hue="Survived", bins=25, kde=True, palette="Set1", alpha=0.6)
plt.title("Age Distribution by Survival Status", fontsize=18, weight="bold")
plt.xlabel("Age", fontsize=13)
plt.ylabel("Frequency", fontsize=13)
plt.legend(title="Survived", labels=["No", "Yes"])
plt.show()
```

```python
sns.set_theme(style="whitegrid")

plt.figure(figsize=(10,6))

sns.boxplot(
    x="Pclass",
    y="Fare",
    data=df,
    palette="Set2",
    linewidth=2,
    fliersize=6
)

plt.title(
    "Ticket Fare Distribution by Passenger Class",
    fontsize=18,
    weight="bold"
)

plt.xlabel("Passenger Class", fontsize=13)
plt.ylabel("Ticket Fare", fontsize=13)
plt.legend(title="Passenger Class", labels=["1st Class", "2nd Class", "3rd Class"])

plt.xticks([0,1,2], ["First", "Second", "Third"])

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

```python
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10,6))  
sns.boxplot(x="Pclass", y="Fare", data=df, palette="Set2", linewidth=2, fliersize=6)
plt.title("Ticket Fare by Passenger Class", fontsize=18, weight="bold")
plt.xlabel("Passenger Class", fontsize=13)
plt.ylabel("Ticket Fare", fontsize=13)
plt.legend(title="Passenger Class", labels=["1st Class", "2nd Class", "3rd Class"]) 
plt.show()
```
